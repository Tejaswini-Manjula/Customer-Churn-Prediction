```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("../data/cleaned_telco_churn.csv")

# Show first 5 rows
print(df.head())

# Dataset information
print(df.info())

# Statistical summary
print(df.describe())

# Missing values
print(df.isnull().sum())
```

```python
sns.countplot(x='Churn', data=df)

plt.title("Customer Churn Distribution")

plt.savefig("../images/churn_distribution.png")

plt.show()
```

# Insight:
# Customers who stayed are more than customers who churned.

```python
sns.countplot(x='gender',
              hue='Churn',
              data=df)

plt.title("Gender vs Churn")

plt.show()
```

# Insight:
# Churn rate is almost similar for male and female customers.

```python
sns.countplot(x='Contract',
              hue='Churn',
              data=df)

plt.title("Contract Type vs Churn")

plt.xticks(rotation=15)

plt.savefig("../images/contract_vs_churn.png")

plt.show()
```

# Insight:
# Customers with month-to-month contracts show higher churn.

```python
sns.histplot(df['MonthlyCharges'],
             kde=True)

plt.title("Monthly Charges Distribution")

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.savefig("../images/heatmap.png")

plt.show()
```

```python
sns.countplot(y='PaymentMethod',
              hue='Churn',
              data=df)

plt.title("Payment Method vs Churn")

plt.savefig("../images/payment_method_vs_churn.png")

plt.show()
```

# Insight:
# Customers using electronic check show higher churn.

```python
sns.countplot(x='MultipleLines',
              hue='Churn',
              data=df)

plt.title("Multiple Lines vs Churn")

plt.xticks(rotation=15)

plt.show()
```

# Insight:
# Customers with multiple lines show moderate churn behavior.

```python
sns.boxplot(x='Churn',
            y='tenure',
            data=df)

plt.title("Tenure vs Churn")

plt.savefig("../images/tenure_vs_churn.png")

plt.show()
```

# Insight:
# Customers with lower tenure are more likely to churn.

# Final Insights

1. Customers with month-to-month contracts showed higher churn.
2. Customers with lower tenure were more likely to leave the service.
3. Monthly charges had an impact on customer churn.
4. Certain payment methods showed higher churn rates.
5. Long-term customers were more likely to stay with the company.
