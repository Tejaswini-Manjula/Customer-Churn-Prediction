```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("../data/cleaned_telco_churn.csv")

# Show first 5 rows
print(df.head())

# Dataset information
print(df.info())

# Statistical summary
print(df.describe())

# Missing values
print(df.isnull().sum())
```

```python
sns.countplot(x='Churn', data=df)

plt.title("Customer Churn Distribution")

plt.savefig("../images/churn_distribution.png")

plt.show()
```

# Insight:
# Customers who stayed are more than customers who churned.

```python
sns.countplot(x='gender',
              hue='Churn',
              data=df)

plt.title("Gender vs Churn")

plt.show()
```

# Insight:
# Churn rate is almost similar for male and female customers.

```python
sns.countplot(x='Contract',
              hue='Churn',
              data=df)

plt.title("Contract Type vs Churn")

plt.xticks(rotation=15)

plt.savefig("../images/contract_vs_churn.png")

plt.show()
```

# Insight:
# Customers with month-to-month contracts show higher churn.

```python
sns.histplot(df['MonthlyCharges'],
             kde=True)

plt.title("Monthly Charges Distribution")

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.savefig("../images/heatmap.png")

plt.show()
```

```python
sns.countplot(y='PaymentMethod',
              hue='Churn',
              data=df)

plt.title("Payment Method vs Churn")

plt.savefig("../images/payment_method_vs_churn.png")

plt.show()
```

# Insight:
# Customers using electronic check show higher churn.

```python
sns.countplot(x='MultipleLines',
              hue='Churn',
              data=df)

plt.title("Multiple Lines vs Churn")

plt.xticks(rotation=15)

plt.show()
```

# Insight:
# Customers with multiple lines show moderate churn behavior.

```python
sns.boxplot(x='Churn',
            y='tenure',
            data=df)

plt.title("Tenure vs Churn")

plt.savefig("../images/tenure_vs_churn.png")

plt.show()
```

# Insight:
# Customers with lower tenure are more likely to churn.

# Final Insights

1. Customers with month-to-month contracts showed higher churn.
2. Customers with lower tenure were more likely to leave the service.
3. Monthly charges had an impact on customer churn.
4. Certain payment methods showed higher churn rates.
5. Long-term customers were more likely to stay with the company.

# Week 6 - Service Based Churn Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/cleaned_telco_churn.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.countplot(x='InternetService', hue='Churn', data=df)
plt.title('Internet Service vs Churn')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='OnlineSecurity', hue='Churn', data=df)
plt.title('Online Security vs Churn')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='TechSupport', hue='Churn', data=df)
plt.title('Tech Support vs Churn')
plt.show()
```

# Week 6 Observation

- Internet service showed differences in customer churn behavior.
- Customers without online security showed higher churn.
- Customers without tech support were more likely to leave.
- Service-related features influence customer retention.

# Week 6 Conclusion

The analysis showed that Internet Service, Online Security, and Tech Support are important factors affecting customer churn. Customers without support-related services were more likely to leave, indicating these services help improve customer retention.

# Week 7 - Customer Retention Recommendations

## Recommendations

1. Focus on customers with month-to-month contracts, as they showed higher churn.

2. Encourage customers to use technical support services to improve retention.

3. Promote online security services to customers who are not using them.

4. Review pricing strategies for customers with higher monthly charges.

5. Provide special offers to new customers with low tenure to reduce early churn.

# Week 7 Observation

- Customer retention strategies can be designed based on these findings.

Based on the analysis, customer churn can be reduced by improving support services, promoting security features, and focusing on customers with short tenure and month-to-month contracts. These insights can help improve customer retention.
